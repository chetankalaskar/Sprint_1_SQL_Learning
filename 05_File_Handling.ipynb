# MODULE 5 - FILE HANDLING

File handling means working with files using Python.

In this module, I learned how to read and write text files, work with CSV and JSON files, store Python objects using Pickle, and work with files and folders using `pathlib`, `os`, and `shutil`.

## 1) READING TEXT FILES

Reading a text file means retrieving the data stored in a file without changing it.

Python uses `open()` in read mode (`"r"`).

### Methods

- `read()` - Reads the entire file
- `readline()` - Reads one line
- `readlines()` - Reads all lines as a list

## 2) WRITING TEXT FILES

Writing stores data into a text file.

### File Modes

- `w` - Write (overwrite)
- `a` - Append (add at the end)
- `x` - Create a new file

```python
sampletxt = open("sample.txt", "w")

sampletxt.write("Hello World.\n")
sampletxt.write("This is Chetan.\n")
sampletxt.write("This file is my module 5 - file handling project.")

sampletxt.close()

print("File created")
```

### Reading a Text File

```python
sampletxt = open("sample.txt", "r")

data = sampletxt.read()

print(data)

sampletxt.close()
```

## 3) CSV FILES

CSV stands for Comma-Separated Values.

CSV stores tabular data. Each value is separated by a comma.

### Uses

- Student records
- Employee data
- Sales reports

### Example

Name,Age,Marks
Ajay,23,99
Kumar,22,95

```python
import pandas as pd

data = {
    "Name": ["Chetan", "Vaishnavi", "Renuka"],
    "Age": [30, 20, 22],
    "City": ["Chennai", "Bangalore", "Hyderabad"]
}

df = pd.DataFrame(data)

df.to_csv("students.csv", index=False)

print("CSV file created.")
```

### Reading a CSV File

```python
import pandas as pd

df = pd.read_csv("students.csv")

print(df)
```

## 4) JSON FILES

JSON stands for JavaScript Object Notation.

JSON stores data in key-value pairs.

It is widely used in APIs and configuration files.

### Example

{
    "Name": "Kiran",
    "Age": 20
}

```python
import json

data = {
    "name": "Chetan",
    "age": 30,
    "city": "Chennai"
}

file = open("student.json", "w")

json.dump(data, file)

file.close()

print("JSON file created.")
```

### Reading a JSON File

```python
import json

file = open("student.json", "r")

data = json.load(file)

file.close()

print(data)
```

## 5) PICKLE FILES

Pickle stores Python objects directly in binary format.

### Terms

- **Serialization** - Saving objects
- **Deserialization** - Loading objects

```python
import pickle

data = {
    "name": "Chetan",
    "age": 30,
    "city": "Chennai"
}

file = open("student.pkl", "wb")

pickle.dump(data, file)

file.close()

print("Pickle file created.")
```

### Reading a Pickle File

```python
import pickle

file = open("student.pkl", "rb")

data = pickle.load(file)

file.close()

print(data)
```

## 6) CONTEXT MANAGER (`with`)

The `with` statement automatically opens and closes the file after use.

It is useful because we do not have to manually close the file.

### Writing a File Using `with`

```python
with open("sample1.txt", "w") as file:
    file.write("Hello World")
    file.write("\nThis is Aishu")

print("File created successfully.")
```

## 7) PATHLIB

`pathlib` provides an object-oriented way to work with file and folder paths.

It is cross-platform and easy to use.

### Create a Path

```python
from pathlib import Path

path = Path("sample.txt")

print(path)
```

### Checking if the File Exists

```python
from pathlib import Path

path = Path("sample.txt")

if path.exists():
    print("File exists")
else:
    print("File does not exist")
```

### Reading a File

```python
from pathlib import Path

path = Path("sample.txt")

data = path.read_text()

print(data)
```

## 9) SHUTIL MODULE

`shutil` stands for Shell Utilities.

It provides high-level file operations.

### Common Functions

- `copy()` - Copy files
- `move()` - Move files
- `rmtree()` - Remove a folder and its contents
- `copytree()` - Copy a complete folder

### Copy a File

```python
import shutil

shutil.copy("sample.txt", "copy_sample.txt")

print("File copied.")
```

## 10) FILE VALIDATION

File validation checks whether:

- a file exists
- a file is readable or writable
- a file is empty
- a file has the correct type

File validation helps to prevent runtime errors.

### Check if a File Exists

```python
import os

if os.path.exists("sample.txt"):
    print("File exists")
else:
    print("File does not exist")
```

### Validating the File Before Reading

```python
import os

filename = "sample.txt"

if os.path.exists(filename):
    file = open(filename, "r")
    print(file.read())
    file.close()
else:
    print("File not found")
```
