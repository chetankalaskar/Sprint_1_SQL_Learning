## 1. Database Design

### Understand

A database is an organized collection of data.

A relational database stores data in tables. A table contains rows and
columns, where each row represents a record and each column represents
a particular attribute.

### Demonstration

For an employee management system, I can create an `Employees` table with
columns such as employee ID, name, department, salary and city.

This demonstrates database design because related employee information is
organized into one structured table.

### My Understanding

I understand database design as organizing data into suitable tables and columns so that the data is easier to manage and retrieve.

```python
import sqlite3

connection = sqlite3.connect("company.db")
cursor = connection.cursor()

print("Database connected successfully.")
```

## 2. Tables, Primary Key and Constraints

### Understand

A table stores related information using rows and columns.

A primary key uniquely identifies each row in a table.

Constraints are rules that control the data stored in a table.

Common constraints include:

- `PRIMARY KEY` - uniquely identifies a record
- `NOT NULL` - prevents a required value from being empty
- `UNIQUE` - prevents duplicate values
- `DEFAULT` - provides a default value

### Demonstration

In my employee table, `employee_id` is the primary key because every employee
should have a unique ID.

The `name` column uses `NOT NULL` because an employee should have a name.

### My Understanding

I understand a primary key as a unique identifier for each record. Constraints
help maintain the quality and consistency of the data.

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS Employees (
    employee_id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    department TEXT,
    salary REAL,
    city TEXT
)
""")

connection.commit()

print("Employees table created successfully.")
```

## 3. Inserting Data

### Understand

The `INSERT INTO` statement is used to add records to a table.

### Demonstration

I will insert sample employee records into the `Employees` table. These records
will be used to practice the SQL queries in the next sections.

### My Understanding

I understand `INSERT INTO` as the SQL command used to add new records to a
table.

```python
employees = [
    (1, "Chetan", "Testing", 50000, "Kalaburagi"),
    (2, "Asha", "Development", 60000, "Pune"),
    (3, "Rahul", "Testing", 55000, "Bengaluru"),
    (4, "Priya", "HR", 45000, "Hyderabad"),
    (5, "Arun", "Development", 65000, "Pune")
]

cursor.executemany("""
INSERT INTO Employees
(employee_id, name, department, salary, city)
VALUES (?, ?, ?, ?, ?)
""", employees)

connection.commit()

print("Employee records inserted successfully.")
```

## 4. SELECT

### Understand

`SELECT` is used to retrieve data from a table.

`SELECT *` means that I want to retrieve all columns.

### Demonstration

I will use `SELECT` to display all employee records from the `Employees` table.

### My Understanding

I learned that SELECT is used when I want to get data from a table.

```python
cursor.execute("SELECT * FROM Employees")

for employee in cursor.fetchall():
    print(employee)
```

## 5. WHERE

### Understand

The `WHERE` clause is used to filter rows based on a condition.

### Demonstration

I can use `WHERE` to retrieve only employees who belong to the Testing
department.

### My Understanding

I learned that WHERE helps me filter the rows and get only the records that match my condition.

```python
cursor.execute("""
SELECT name, department, salary
FROM Employees
WHERE department = 'Testing'
""")

for employee in cursor.fetchall():
    print(employee)
```

## 6. ORDER BY

### Understand

`ORDER BY` is used to sort the result of a query.

`ASC` sorts in ascending order and `DESC` sorts in descending order.

### Demonstration

I will sort employees according to their salary from highest to lowest.

### My Understanding

I learned that ORDER BY is used to arrange the query results, such as sorting salary from highest to lowest.

```python
cursor.execute("""
SELECT name, salary
FROM Employees
ORDER BY salary DESC
""")

for employee in cursor.fetchall():
    print(employee)
```

## 7. DISTINCT and LIMIT

### DISTINCT

`DISTINCT` removes duplicate values from the result.

For example, more than one employee can belong to the same city. `DISTINCT`
allows me to display each city only once.

### LIMIT

`LIMIT` restricts the number of rows returned by a query.

### My Understanding

I understand `DISTINCT` as a way to remove duplicate results and `LIMIT` as
a way to control how many records are returned.

```python
print("Unique cities:")

cursor.execute("""
SELECT DISTINCT city
FROM Employees
""")

for city in cursor.fetchall():
    print(city[0])

print("\nFirst 3 employees:")

cursor.execute("""
SELECT *
FROM Employees
LIMIT 3
""")

for employee in cursor.fetchall():
    print(employee)
```

## 8. GROUP BY and HAVING

### GROUP BY

`GROUP BY` groups rows that have the same value.

For example, I can group employees by department and count the employees
in each department.

### HAVING

`HAVING` filters grouped results.

### Demonstration

I will group employees by department and then display only departments that
have more than one employee.

### My Understanding

I learned that GROUP BY creates groups based on a column, such as department. HAVING can then filter those groups based on a condition.

```python
cursor.execute("""
SELECT department, COUNT(*) AS employee_count
FROM Employees
GROUP BY department
HAVING COUNT(*) > 1
""")

for result in cursor.fetchall():
    print(result)
```
