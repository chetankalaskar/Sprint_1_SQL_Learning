## 1. Database Design

### Understand

A database is an organized collection of data.

a particular attribute.

```python
import sqlite3

connection = sqlite3.connect("company.db")
cursor = connection.cursor()

print("Database connected successfully.")
```

## 2. Tables, Primary Key and Constraints

### Understand

A table stores related information using rows and columns.

A primary key uniquely identifies each row in a table.

Constraints are rules that control the data stored in a table.

Common constraints include:

- `PRIMARY KEY` - uniquely identifies a record
- `NOT NULL` - prevents a required value from being empty
- `UNIQUE` - prevents duplicate values
- `DEFAULT` - provides a default value

### Demonstration

In my employee table, `employee_id` is the primary key because every employee
should have a unique ID.

The `name` column uses `NOT NULL` because an employee should have a name.

### My Understanding

I understand a primary key as a unique identifier for each record. Constraints
help maintain the quality and consistency of the data.

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS Employees (
    employee_id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    department TEXT,
    salary REAL,
    city TEXT
)
""")

connection.commit()

print("Employees table created successfully.")
```

## 3. Foreign Key

### Understand

A `FOREIGN KEY` is a column that creates a connection between two tables.

It refers to the `PRIMARY KEY` of another table.

A foreign key helps maintain a valid relationship between related tables.

### Demonstration

In my database, the `Departments` table has `department_id` as its primary key.

The `Employees2` table has `department_id` as a foreign key that refers to the `department_id` in the `Departments` table.

### My Understanding

I understand a foreign key as a column that connects two tables by referring to the primary key of another table.

## 4. Other Constraints

### UNIQUE

`UNIQUE` ensures that all values in a column are different.

For example, an employee email address should normally be unique.

### CHECK

`CHECK` ensures that the value entered into a column satisfies a specified condition.

For example, salary can be checked to ensure that it is greater than or equal to zero.

### DEFAULT

`DEFAULT` provides a value automatically when no value is supplied.

For example, if no city is provided, the default city can be set to `Hyderabad`.

### My Understanding

I understand that constraints are rules used to control the type and validity of data stored in a table.

```python
cursor.execute("DROP TABLE IF EXISTS EmployeeDetails")

cursor.execute("""
CREATE TABLE EmployeeDetails (
    employee_id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    email TEXT UNIQUE,
    salary REAL CHECK (salary >= 0),
    city TEXT DEFAULT 'Hyderabad'
)
""")

connection.commit()

print("EmployeeDetails table created successfully.")
```

## 8. WHERE

### Understand

The `WHERE` clause is used to filter rows based on a condition.

### Demonstration

I can use `WHERE` to retrieve only employees who belong to the Testing

### My Understanding

I learned that WHERE helps me filter the rows and get only the records that match my condition.

```python
cursor.execute("""
SELECT name, department, salary
FROM Employees
WHERE department = 'Testing'
""")

for employee in cursor.fetchall():
    print(employee)
```

## 9. ORDER BY

### Understand

`ORDER BY` is used to sort the result of a query.

`ASC` sorts in ascending order and `DESC` sorts in descending order.

### Demonstration

I will sort employees according to their salary from highest to lowest.

### My Understanding

I learned that ORDER BY is used to arrange the query results, such as sorting salary from highest to lowest.

```python
cursor.execute("""
SELECT name, salary
FROM Employees
ORDER BY salary DESC
""")

for employee in cursor.fetchall():
    print(employee)
```

## 10. DISTINCT and LIMIT

### DISTINCT

`DISTINCT` removes duplicate values from the result.

For example, more than one employee can belong to the same city. `DISTINCT`
allows me to display each city only once.

### LIMIT

`LIMIT` restricts the number of rows returned by a query.

### My Understanding

I understand `DISTINCT` as a way to remove duplicate results and `LIMIT` as
a way to control how many records are returned.

```python
print("Unique cities:")

cursor.execute("""
SELECT DISTINCT city
FROM Employees
""")

for city in cursor.fetchall():
    print(city[0])

print("\nFirst 3 employees:")

cursor.execute("""
SELECT *
FROM Employees
LIMIT 3
""")

for employee in cursor.fetchall():
    print(employee)
```

## 11. GROUP BY and HAVING

### GROUP BY

`GROUP BY` groups rows that have the same value.

For example, I can group employees by department and count the employees
in each department.

### HAVING

`HAVING` filters grouped results.

### Demonstration

I will group employees by department and then display only departments that
have more than one employee.

### My Understanding

I learned that GROUP BY creates groups based on a column, such as department. HAVING can then filter those groups based on a condition.

```python
cursor.execute("""
SELECT department, COUNT(*) AS employee_count
FROM Employees
GROUP BY department
HAVING COUNT(*) > 1
""")

for result in cursor.fetchall():
    print(result)
```

## 1. SQL JOIN

### Understand

A `JOIN` is used to combine data from two or more tables.

Usually, the tables have a common column that can be used to connect them.

For example, an employee can have a `department_id`, and the department table can also have a `department_id`. I can use this column to connect the two tables.

### Demonstration

I will create two tables for my example:

`Departments` stores department information.
`Employees2` stores employee information.

The `department_id` column will connect the two tables.

### My Understanding

I understand a JOIN as a way to bring related information from different tables into one result.

## 2. Creating Tables for JOIN Practice

### Understand

To practice joins, I need two tables that have a related column.

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS Departments (
    department_id INTEGER PRIMARY KEY,
    department_name TEXT NOT NULL
)
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS Employees2 (
    employee_id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    department_id INTEGER,
    FOREIGN KEY (department_id) REFERENCES Departments(department_id)
)
""")

connection.commit()

print("Tables for JOIN practice created successfully.")
```

```python
departments = [
    (1, "Testing"),
    (2, "Development"),
    (3, "HR")
]

employees2 = [
    (1, "Chetan", 1),
    (2, "Asha", 2),
    (3, "Rahul", 1),
    (4, "Priya", 3),
    (5, "Arun", 2)
]

cursor.executemany("""
INSERT OR IGNORE INTO Departments
(department_id, department_name)
VALUES (?, ?)
""", departments)

cursor.executemany("""
INSERT OR IGNORE INTO Employees2
(employee_id, name, department_id)
VALUES (?, ?, ?)
""", employees2)

connection.commit()

print("Sample data inserted successfully.")
```

### 3. INNER JOIN

### Understand

`INNER JOIN` returns only the rows that have matching values in both tables.

### Demonstration

I will connect `Employees2` and `Departments` using `department_id`.

This allows me to see the employee name together with the department name.

### My Understanding

I understand `INNER JOIN` as showing only the records where both tables have a matching value.

```python
cursor.execute("""
SELECT Employees2.name, Departments.department_name
FROM Employees2
INNER JOIN Departments
ON Employees2.department_id = Departments.department_id
""")

for result in cursor.fetchall():
    print(result)
```

## 4. LEFT JOIN
   
### Understand

`LEFT JOIN` returns all the rows from the left table and the matching rows from the right table.

If there is no match, the right-side information will be `NULL`.

### Demonstration

Here, `Employees2` is the left table. Therefore, all employees will be included in the result.

### My Understanding

I understand `LEFT JOIN` as keeping all the records from the first table and adding matching information from the second table.

```python
cursor.execute("""
SELECT Employees2.name, Departments.department_name
FROM Employees2
LEFT JOIN Departments
ON Employees2.department_id = Departments.department_id
""")

for result in cursor.fetchall():
    print(result)
```

## 5. RIGHT JOIN
   
### Understand

`RIGHT JOIN` returns all rows from the right table and the matching rows from the left table.

If there is no match, the left-side information will be `NULL`.

### Demonstration

If `Departments` is the right table, a `RIGHT JOIN` makes sure all departments are included in the result.

### My Understanding

I understand `RIGHT JOIN` as keeping all the records from the second table and adding matching records from the first table.

```python
cursor.execute("""
SELECT Employees2.name, Departments.department_name
FROM Employees2
RIGHT JOIN Departments
ON Employees2.department_id = Departments.department_id
""")

for result in cursor.fetchall():
    print(result)
```

## 6. FULL JOIN
   
### Understand

`FULL JOIN` returns all rows from both tables.

If there is a matching record, the information is combined. If there is no match, `NULL` is shown for the missing side.

### Demonstration

A `FULL JOIN` is useful when I want to see all records from both tables, whether they have a match or not.

### My Understanding

I understand `FULL JOIN` as keeping all the records from both tables and matching them wherever possible.

```python
cursor.execute("""
SELECT Employees2.name, Departments.department_name
FROM Employees2
FULL JOIN Departments
ON Employees2.department_id = Departments.department_id
""")

for result in cursor.fetchall():
    print(result)
```

## 7. CROSS JOIN

### Understand

`CROSS JOIN` combines every row from the first table with every row from the second table.

This creates every possible combination between the two tables.

### Demonstration

If I have 5 employees and 3 departments, the result will contain 15 combinations.

### My Understanding

I understand `CROSS JOIN` as combining every row of one table with every row of another table.

```python
cursor.execute("""
SELECT Employees2.name, Departments.department_name
FROM Employees2
CROSS JOIN Departments
""")

for result in cursor.fetchall():
    print(result)
```

## 1. UNION

### Understand

`UNION` is used to combine the results of two or more `SELECT` queries into a single result.

`UNION` removes duplicate rows from the final result.

The queries used with `UNION` should have the same number of columns and compatible data types.

### Demonstration

I will create two tables containing employee names and combine their results using `UNION`.

### My Understanding

I understand `UNION` as a way to combine the results of multiple `SELECT` queries while removing duplicate records.

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS Developers (
    employee_id INTEGER,
    name TEXT
)
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS Testers (
    employee_id INTEGER,
    name TEXT
)
""")

developers = [
    (1, "Chetan"),
    (2, "Asha"),
    (3, "Rahul")
]

testers = [
    (4, "Priya"),
    (5, "Arun"),
    (6, "Rahul")
]

cursor.executemany(
    "INSERT INTO Developers VALUES (?, ?)",
    developers
)

cursor.executemany(
    "INSERT INTO Testers VALUES (?, ?)",
    testers
)

connection.commit()

print("Tables and data created successfully.")
```

```python
cursor.execute("""
SELECT name
FROM Developers

UNION

SELECT name
FROM Testers
""")

for result in cursor.fetchall():
    print(result)
```

### My Understanding

`UNION` combines the results of multiple `SELECT` queries and removes duplicate values.

In this example, `Rahul` appears in both tables but is displayed only once.

## 2. CASE

### Understand

`CASE` is used to apply conditions and return different values based on those conditions.

It works similar to `if-elif-else` logic in programming.

### Demonstration

I will classify employees based on their salary.

- Salary >= 60000 → High Salary
- Salary >= 50000 → Medium Salary
- Otherwise → Low Salary

### My Understanding

I understand `CASE` as a way to apply conditional logic inside an SQL query.

```python
cursor.execute("""
SELECT
    name,
    salary,
    CASE
        WHEN salary >= 60000 THEN 'High Salary'
        WHEN salary >= 50000 THEN 'Medium Salary'
        ELSE 'Low Salary'
    END AS salary_category
FROM Employees
""")

for result in cursor.fetchall():
    print(result)
```

## 3. Subqueries

### Understand

A subquery is a query written inside another SQL query.

The inner query executes first, and its result is used by the outer query.

### Demonstration

I will find the employee who has the highest salary using a subquery.

### My Understanding

I understand a subquery as a query written inside another query that helps me use the result of one query in another query.

```python
cursor.execute("""
SELECT name, salary
FROM Employees
WHERE salary = (
    SELECT MAX(salary)
    FROM Employees
)
""")

for result in cursor.fetchall():
    print(result)
```

### Another Example

I can also use a subquery to find employees whose salary is greater than the average salary.

```python
cursor.execute("""
SELECT name, salary
FROM Employees
WHERE salary > (
    SELECT AVG(salary)
    FROM Employees
)
""")

for result in cursor.fetchall():
    print(result)
```

## 4. Common Table Expressions (CTEs)

### Understand

A Common Table Expression, or CTE, is a temporary result set created using the `WITH` statement.

CTEs make complex queries easier to read and understand.

### Demonstration

I will create a CTE containing employees whose salary is greater than 50000.

### My Understanding

I understand a CTE as a temporary named result that I can use inside a query to make complex SQL easier to write and read.

```python
cursor.execute("""
WITH HighSalaryEmployees AS (
    SELECT name, salary
    FROM Employees
    WHERE salary > 50000
)

SELECT *
FROM HighSalaryEmployees
""")

for result in cursor.fetchall():
    print(result)
```

## 5. Views

### Understand

A view is a virtual table created from the result of an SQL query.

A view is useful when I frequently need the same query result.

### Demonstration

I will create a view containing employees whose salary is greater than 50000.

### My Understanding

I understand a view as a virtual table that allows me to save and reuse a SQL query.

```python
cursor.execute("DROP VIEW IF EXISTS HighSalaryEmployees")

cursor.execute("""
CREATE VIEW HighSalaryEmployees AS
SELECT name, department, salary
FROM Employees
WHERE salary > 50000
""")

connection.commit()

print("View created successfully.")
```

```python
cursor.execute("""
SELECT *
FROM HighSalaryEmployees
""")

for result in cursor.fetchall():
    print(result)
```

## 6. Indexes

## 7. Window Functions

### Understand

Window functions perform calculations across a set of rows without grouping those rows into a single result.

Unlike `GROUP BY`, window functions keep every individual row in the result.

### Demonstration

I will use a window function to rank employees according to their salary.

### My Understanding

I understand window functions as functions that allow me to perform calculations across rows while still keeping each individual record.

```python
cursor.execute("""
SELECT
    name,
    salary,
    RANK() OVER (ORDER BY salary DESC) AS salary_rank
FROM Employees
""")

for result in cursor.fetchall():
    print(result)
```

## 8. ROW_NUMBER()

### Understand

`ROW_NUMBER()` assigns a unique sequential number to every row.

Even if two employees have the same salary, they receive different row numbers.

### Demonstration

I will create a student table and assign a row number based on marks from highest to lowest.

### My Understanding

I understand `ROW_NUMBER()` as assigning a unique sequential number to every row.

```python
cursor.execute("DROP TABLE IF EXISTS StudentMarks")

cursor.execute("""
CREATE TABLE StudentMarks (
    student_id INTEGER PRIMARY KEY,
    name TEXT,
    marks INTEGER
)
""")

students = [
    (1, "Chetan", 90),
    (2, "Asha", 80),
    (3, "Rahul", 80),
    (4, "Priya", 70),
    (5, "Arun", 60)
]

cursor.executemany("""
INSERT INTO StudentMarks
(student_id, name, marks)
VALUES (?, ?, ?)
""", students)

connection.commit()

print("Student marks inserted successfully.")
```

```python
cursor.execute("""
SELECT
    name,
    marks,
    ROW_NUMBER() OVER (ORDER BY marks DESC) AS row_num
FROM StudentMarks
""")

for result in cursor.fetchall():
    print(result)
```

## 9. RANK()

### Understand

`RANK()` assigns a rank to each row based on the specified ordering.

If two rows have the same value, they receive the same rank.

When a tie occurs, the next rank is skipped.

For example:

Marks | Rank
--- | ---
90 | 1
80 | 2
80 | 2
70 | 4

### My Understanding

I understand `RANK()` as assigning the same rank to tied values and skipping the next rank.

```python
cursor.execute("""
SELECT
    name,
    marks,
    RANK() OVER (ORDER BY marks DESC) AS student_rank
FROM StudentMarks
""")

for result in cursor.fetchall():
    print(result)
```

## 10. DENSE_RANK()

### Understand

`DENSE_RANK()` is similar to `RANK()`.

The difference is that when there is a tie, the next rank is not skipped.

For example:

Marks | Dense Rank
--- | ---
90 | 1
80 | 2
80 | 2
70 | 3

### My Understanding

I understand `DENSE_RANK()` as assigning the same rank to tied values without leaving gaps in the ranking.

```python
cursor.execute("""
SELECT
    name,
    marks,
    DENSE_RANK() OVER (ORDER BY marks DESC) AS dense_rank
FROM StudentMarks
""")

for result in cursor.fetchall():
    print(result)
```

## 11. LAG()

### Understand

`LAG()` returns the value from the previous row.

It is useful when I want to compare the current row with the previous row.

### Demonstration

I will display each student's marks along with the marks of the previous student.

### My Understanding

I understand `LAG()` as a function that allows me to access a value from the previous row.

```python
cursor.execute("""
SELECT
    name,
    marks,
    LAG(marks) OVER (ORDER BY marks DESC) AS previous_marks
FROM StudentMarks
""")

for result in cursor.fetchall():
    print(result)
```

## 12. LEAD()

### Understand

`LEAD()` returns the value from the next row.

It is useful when I want to compare the current row with the next row.

### Demonstration

I will display each student's marks along with the marks of the next student.

### My Understanding

I understand `LEAD()` as a function that allows me to access a value from the next row.

```python
cursor.execute("""
SELECT
    name,
    marks,
    LEAD(marks) OVER (ORDER BY marks DESC) AS next_marks
FROM StudentMarks
""")

for result in cursor.fetchall():
    print(result)
```

# Window Functions Comparison

### ROW_NUMBER()

Every row gets a unique number.

90 → 1  
80 → 2  
80 → 3  
70 → 4  

### RANK()

Tied rows get the same rank and the next rank is skipped.

90 → 1  
80 → 2  
80 → 2  
70 → 4  

### DENSE_RANK()

Tied rows get the same rank but no rank is skipped.

90 → 1  
80 → 2  
80 → 2  
70 → 3  

# LAG vs LEAD

| Function | Purpose |
|----------|---------|
| `LAG()` | Gets the value from the previous row |
| `LEAD()` | Gets the value from the next row |
