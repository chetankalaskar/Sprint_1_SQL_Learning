## 1. Print Statements

The `print()` function is used to display information on the screen.

It is one of the simplest and most commonly used functions in Python.

```python
# Print Statement

print("Hello, Stackly!")
print("Welcome to Python Fundamentals.")
print("This is my Day 1 Python practice.")
```

### My Understanding

I understood that `print()` is used when I want Python to show something as output.
I used it to display text on the screen. Different `print()` statements display their
messages one after another.

## 2. Variables

A variable is a name used to store a value.

Python allows us to store different types of information in variables.

For example:
- A name can be stored as text.
- An age can be stored as an integer.
- A height can be stored as a decimal number.

```python
# Variables

name = "Chetan"
age = 25
height = 5.8
city = "Kalaburagi"

print("Name:", name)
print("Age:", age)
print("Height:", height)
print("City:", city)
```

### My Understanding

I understood a variable as a name that refers to a value stored in the program.
For example, I used `name` to store text and `age` to store a number.
When I use the variable later, Python gives me the value stored in it.

## 3. Data Types

A data type tells Python what kind of value is stored in a variable.

Some basic Python data types are:

| Data Type | Description | Example |
|---|---|---|
| `str` | Text | `"Chetan"` |
| `int` | Whole number | `25` |
| `float` | Decimal number | `5.8` |
| `bool` | True or False | `True` |

The `type()` function can be used to check the data type of a value.

```python
# Checking Data Types

name = "Chetan"
age = 25
height = 5.8
is_student = True

print("Name type:", type(name))
print("Age type:", type(age))
print("Height type:", type(height))
print("Student type:", type(is_student))
```

### My Understanding

I understood that different values have different data types.
For example, a person's name is text (`str`), age is a whole number (`int`),
and height can be a decimal number (`float`).

I used `type()` to check what type of value Python stored in each variable.

## 4. Arithmetic Operators

Arithmetic operators are used to perform mathematical calculations.

The main arithmetic operators are:

- `+` Addition
- `-` Subtraction
- `*` Multiplication
- `/` Division
- `//` Floor division
- `%` Modulus or remainder
- `**` Power

```python
# Arithmetic Operators

a = 10
b = 3

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Remainder:", a % b)
print("Power:", a ** b)
```

### My Understanding

I used arithmetic operators to perform calculations in Python.
For example, `+` adds two numbers and `-` subtracts them.
I also noticed that `/` gives a division result with decimals, while `//`
gives floor division. `%` gives the remainder.

## 5. User Input

The `input()` function is used to take information from the user.

By default, `input()` returns the entered value as text.

We can use `int()` to convert input into an integer and `float()` to convert input into a decimal number.

```python
# Taking User Input

name = input("Enter your name: ")
age = int(input("Enter your age: "))

print("Hello", name)
print("Your age is", age)
```

### My Understanding

I understood that `input()` allows the program to receive information from the user.
The value received from `input()` is treated as text by default.

Because age is a number, I used `int()` to convert the entered text into an integer.

## 6. Simple Calculator

Now I will combine variables, user input, data conversion, and arithmetic operators to create a simple calculator.

The user enters two numbers, and Python performs basic mathematical operations on them.

```python
# Simple Calculator

num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

print("Addition:", num1 + num2)
print("Subtraction:", num1 - num2)
print("Multiplication:", num1 * num2)
print("Division:", num1 / num2)
```

### My Understanding

I created this calculator to combine the concepts I learned today.
I used `input()` to get two numbers from the user, `float()` to convert them
into numbers, and arithmetic operators to perform the calculations.

This example helped me understand how different basic Python concepts can
work together in one program.
