# Module 1 - Python Fundamentals

### Variables

A variable is a named place used to store a value.

It is like a box with a label which keeps some information. The value inside the variable can be changed whenever we want.

Example:

`age = 30`

Here:

- `age` → variable
- `30` → value stored in the variable

We can use `age` instead of writing the number `30` everywhere.

### Data Types

A data type tells Python what type of information is stored in a variable.

Different types of information need different data types. It tells whether the value is a number, text, decimal number or True/False.

Some common data types are:

- Integer (`int`) → Whole numbers
- Float (`float`) → Decimal numbers
- String (`str`) → Text
- Boolean (`bool`) → True or False

### Type Casting

Type casting means changing one data type into another.

It is used when we want to change data from one form to another, like text to a number.

Example:

`"25"` → `str`

`25` → `int`

Both represent twenty-five, but Python treats them as different data types.

### Operators

Operators are symbols which are used to perform actions on values.

They are mainly used for calculations, comparisons and checking conditions.

Some examples are:

Arithmetic Operators:

- `+` → Add
- `-` → Subtract
- `*` → Multiply
- `/` → Divide

Comparison Operators:

- `>` → Greater than
- `<` → Less than
- `==` → Equal to

### Input and Output

Input means giving information to the program.

Output means the program shows the result back to us.

Almost every program takes some input, processes it and produces an output.

Example:

`a = 5`

`b = 3`

`print(a + b)`

- Input → `5, 3`
- Output → `8`

The program takes the input 5 and 3, performs addition and gives the output 8.

### Comments

Comments are notes written inside the program.

Comments help us and other programmers understand what the code is doing.

Python ignores comments while running the code.

### Single Line Comment

A single line comment uses the `#` symbol.

### Docstring

A docstring is generally written using triple quotes `"""` at the beginning and end.

It is mainly used to describe a function, class or module.

### if / elif / else

`if`, `elif` and `else` are used when Python needs to make decisions.

- `if` checks the first condition.
- `elif` checks another condition if the first condition is false.
- `else` runs when none of the conditions are true.

Only one matching block is executed.

### Example

Imagine traffic lights.

- If the light is Green → Go
- Else if the light is Yellow → Slow
- Else → Stop

Python checks the conditions one by one.

As soon as one condition is true, it performs that action.

```python
# Odd or Even

num = int(input("Enter a number: "))

if num % 2 == 0:
    print("Even Number")
else:
    print("Odd Number")
```

### for loop

#### Definition

A `for` loop is a control statement used to repeat a block of code for each item in a sequence or for a specific number of times.

### Real-Life Example

Imagine a teacher checking attendance for 30 students.

Instead of checking each student's name manually, the teacher goes through the class list one student at a time until everyone is checked.

A `for` loop works in a similar way. It repeats the same task for every item in a sequence.

### Example

```python
# Prime Number

num = int(input("Enter a number: "))

count = 0

for i in range(1, num + 1):
    if num % i == 0:
        count += 1

if count == 2:
    print(num, "is a prime number")
else:
    print(num, "is not a prime number")
```

### while loop

#### Definition

A `while` loop is a control statement that repeats a block of code as long as a specified condition is `True`.

### Real-Life Example

Imagine filling a water tank.

While the tank is not full, keep filling it.

When the tank becomes full, stop.

The loop continues until the condition becomes `False`.

### Example

```python
# Sum of first 10 numbers

i = 1
total = 0

while i <= 10:
    total += i
    i += 1

print("Sum =", total)
```

## Break

### Definition

The `break` statement is used to immediately stop a loop, even if the loop has not completed all its iterations.

### Real-Life Example

Imagine you're searching for your friend in a crowd.

As soon as you find them, you stop searching.

You don't continue looking through the remaining people.

That's exactly what `break` does.

### Example

```python
# Find a number in a list

numbers = [10, 20, 30, 40, 50]

search = int(input("Enter number to search: "))

for num in numbers:
    if num == search:
        print("Number found")
        break
else:
    print("Not in the list")
```

# Continue

## Definition

The `continue` statement is used to skip the current iteration of a loop and continue with the next iteration.

## Real-Life Example

Imagine a teacher checking homework.

One student forgot the homework.

The teacher skips that student and continues checking the rest of the class.

The checking doesn't stop. It simply skips one student.

### Example

```python
# Skip negative numbers in the list

lst = [10, -25, 40, -57, 89, 90, -98]

for number in lst:
    if number < 0:
        continue
    print(number)
```

# Pass

### Definition

The `pass` statement is a placeholder that does nothing.

It is used when Python requires a statement, but no action needs to be performed.

### Real-Life Example

Imagine your teacher tells you:

"Leave this question blank for now. We'll come back to it later."

You haven't answered it, but you also haven't made a mistake.

That's what `pass` does. It tells Python, "Do nothing here for now."

### Example

```python
# Skip a condition

num = int(input("Enter a number: "))

if num > 0:
    pass
else:
    print("Number is not positive")

print("Program completed")
```

# List

### Definition

A list is an ordered and mutable collection that can store multiple items.

It allows duplicate values and the items can be added, removed or modified.

- Ordered → Items keep their position.
- Mutable → Items can be changed after creation.

### Real-Life Example

Think of your shopping list.

Milk  
Bread  
Eggs  
Rice

You can:

- add a new item
- remove an item
- change an item

Just like a Python list.

### Example

`fruits = ["Apple", "Banana", "Mango"]`

```python
# Create a List (Adding names of the students)

students = ["Renu", "Yogita", "Vaishu", "Chetan"]

print(students)
```

```python
# Add a New Student (append())

students = ["Renu", "Yogita", "Vaishu", "Chetan"]

students.append("Arthi")

print(students)
```

```python
# Remove a Student (remove())

students = ["Renu", "Yogita", "Vaishu", "Chetan"]

students.remove("Chetan")

print(students)
```

```python
# Change a name of Student (using Index)

students = ["Renu", "Yogita", "Vaishu", "Chetan"]

students[3] = "Chetu"

print(students)
```

```python
# Call a particular Student

students = ["Renu", "Yogita", "Vaishu", "Chetan"]

print(students[2])
```

```python
# Find the Total Number of Students

students = ["Renu", "Yogita", "Vaishu", "Chetan"]

print(len(students))
```

```python
# Sort in order (Alphabetical order)

students = ["Renu", "Yogita", "Vaishu", "Chetan"]

students.sort()

print(students)
```

## Tuple

### Definition

A tuple is an ordered and immutable collection that stores multiple items.

Once a tuple is created, its values cannot be changed.

- Ordered → Items keep their position.
- Immutable → Items cannot be modified.

### Real-Life Example

Think of your Date of Birth.

Example:

15-08-2000

Your date of birth doesn't change.

Similarly, once a tuple is created, its values remain the same.

### Easy way to remember

- List (`[]`) → Can add, remove and change items.
- Tuple (`()`) → Can store and access items, but cannot be modified after creation.

### Example

`birth_date = (15, 8, 2000)`

```python
# Create tuple and access elements

fruits = ("jack fruit", "gauva", "kiwi")

print(fruits)
print(fruits[0])
print(fruits[2])
```

```python
# Negative indexing

fruits = ("apple", "banana", "mango")

print(fruits[-1])
```

```python
# Slicing

fruits = ("apple", "banana", "mango", "orange", "papaya")

print(fruits[1:4])
```

```python
# Tuple length

fruits = ("apple", "banana", "mango", "orange", "papaya")

print(len(fruits))
```

```python
# Membership

fruits = ("apple", "banana", "mango", "orange", "papaya")

print("apple" in fruits)
print("lemon" in fruits)
```

```python
# Concatenation

fruit = ("apple", "orange", "banana")
colour = ("red", "orange", "yellow")

print(fruit + colour)
```

```python
# Tuple unpacking

stdnt = ("Vaishu", 20, "Latur")

name, age, city = stdnt

print(name)
print(age)
print(city)
```

```python
# Count elements

elements = ("apple", "orange", "banana", "red", "orange", "yellow")

print(elements.count("orange"))
```

```python
# Nested tuple

fruitcolor = (
    ("apple", "red"),
    ("banana", "yellow"),
    ("Guava", "green")
)

print(fruitcolor[1])
print(fruitcolor[2][0])
```

```python
# Immutable

fruits = ("apple", "banana", "mango")

fruits[1] = "orange"
```

## Set

### Definition

A set is an unordered collection of unique items.

It automatically removes duplicate values.

- Unordered → Items do not have a fixed position.
- Unique → Duplicate values are not allowed.

### Real-Life Example

Imagine a classroom.

Five students write their favourite fruit:

Apple  
Mango  
Apple  
Orange  
Mango

If the teacher wants only the unique fruits, the result becomes:

Mango  
Orange

This is how a set works.

### Example

```python
# Only unique items

numbers = {1, 2, 2, 3, 4, 4}

print(numbers)
```

```python
# Adding items

fruits = {"apple", "banana"}

fruits.add("mango")

print(fruits)
```

```python
# Removing items

fruits = {"apple", "banana", "mango"}

fruits.remove("banana")

print(fruits)
```

```python
# Intersection of 2 sets

frt = {"apple", "orange", "banana"}
clr = {"red", "orange", "green"}

print(frt.intersection(clr))
```

## List Comprehension

### Definition

List comprehension is a short and simple way to create a new list.

It allows us to write the loop and optional condition in a single line.

### Real-Life Example

Imagine a teacher writing only the names of students who passed an exam.

Instead of writing a long process, the teacher directly creates a new list containing only the passed students.

Similarly, list comprehension creates a new list in one line.

```python
words = ["cat", "elephant", "dog"]

lengths = [len(word) for word in words]

print(lengths)
```

## Dictionary Comprehension

### Definition

Dictionary comprehension is a short and simple way to create a new dictionary.

It allows us to create key-value pairs using a loop and optional condition in a single line.

### Real-Life Example

Imagine a teacher preparing a report card.

Instead of writing each student's name and mark one by one, the teacher quickly creates the entire report.

Similarly, dictionary comprehension creates key-value pairs efficiently.

```python
fruits = ["apple", "banana", "mango"]

fruit_length = {fruit: len(fruit) for fruit in fruits}

print(fruit_length)
```

## Lambda Functions

### Definition

A lambda function is a small anonymous function written in a single line.

It is commonly used for simple operations where we don't need to create a full function using `def`.

## Variable-Length Arguments (`*args`, `**kwargs`)

### Definition

Variable-length arguments allow a function to accept any number of arguments.

`*args` collects multiple positional arguments into a tuple.

`**kwargs` collects multiple keyword arguments into a dictionary.

### Real-Life Example

Imagine a fruit basket.

Some days it contains:

Apple  
Mango

Other days it contains:

Apple  
Mango  
Banana  
Orange

The basket can hold any number of fruits.

Similarly, `*args` and `**kwargs` allow a function to receive any number of inputs.

#### Example using `*args`

```python
def largest(*numbers):
    print("Largest =", max(numbers))

largest(10, 50, 20, 40)
```

#### Example using `**kwargs`

```python
def details(**data):
    for key, value in data.items():
        print(key, ":", value)

details(name="Chetan", age=30, city="Bangalore")
```

## Recursive Functions

### Definition

A recursive function is a function that calls itself to solve a problem.

It continues calling itself until a stopping condition, called the base case, is reached.

### Real-Life Example

Imagine standing between two mirrors.

You see your reflection repeating again and again.

Similarly, a recursive function keeps calling itself until it reaches the stopping condition.

### Example

```python
# Factorial of a number

def factorial(n):
    if n == 1:
        return 1
    return n * factorial(n - 1)

print(factorial(5))
```

## map()

### Definition

The `map()` function applies a given function to every item in an iterable such as a list or tuple.

It returns a map object, which can be converted into a list if required.

### Real-Life Example

Imagine a teacher adds 5 bonus marks to every student's score.

Instead of updating each score one by one, the teacher applies the same rule to all students.

That's what `map()` does.

```python
fruits = ["apple", "banana", "mango"]

result = list(map(str.upper, fruits))

print(result)
```

## filter()

### Definition

The `filter()` function selects only the items that satisfy a given condition.

It returns a filter object, which can be converted into a list if required.

### Real-Life Example

Imagine a security guard allows only adults into a movie.

Children are filtered out.

Similarly, `filter()` keeps only the items that meet the condition.

```python
fruits = ["apple", "banana", "kiwi", "orange"]

result = list(filter(lambda x: len(x) > 5, fruits))

print(result)
```

## any()

### Definition

The `any()` function returns `True` if at least one element in an iterable is `True`.

If none of the elements are `True`, it returns `False`.

### Real-Life Example

Imagine checking whether at least one light in a room is ON.

If one light is ON, the answer is Yes.

That's how `any()` works.

```python
numbers = [1, 3, 5, 8]

result = any(num % 2 == 0 for num in numbers)

print(result)

# Since 8 is even, any() returns True.
```

## all()

### Definition

The `all()` function returns `True` only if every element in an iterable is `True`.

If even one element is `False`, it returns `False`.

### Real-Life Example

Imagine checking whether all students submitted their assignments.

If even one student hasn't submitted, the answer is No.

That's how `all()` works.

```python
numbers = [1, 3, 6, 8]

result = all(num % 2 == 0 for num in numbers)

print(result)

# Since 1 and 3 are odd, all() returns False.
# all() returns True only when all the numbers satisfy the condition.
```

## Generators

### Definition

A generator is a special type of function that produces values one at a time instead of returning all the values at once.

### Real-Life Example

Imagine a ticket machine.

It gives one ticket each time you press the button instead of printing all the tickets at once.

A generator works in a similar way.

```python
def square():
    for i in range(2, 6):
        yield i * i

gen = square()

for num in gen:
    print(num)
```

## yield

### Definition

The `yield` keyword is used inside a generator function to return one value at a time.

It also saves the current state of the function, so the function can continue from where it stopped when called again.

### Real-Life Example

Think of a TV series.

You watch one episode today and continue with the next episode tomorrow.

It doesn't start again from Episode 1 every time.

Similarly, `yield` pauses the function and allows it to continue later.

```python
def even():
    for i in range(1, 11):
        if i % 2 == 0:
            yield i

for num in even():
    print(num)
```

## Decorators

### Definition

A decorator is a function that adds extra functionality to another function without changing its original code.

### Real-Life Example

Think of gift wrapping.

The gift remains the same, but wrapping adds something extra to it.

Similarly, a decorator adds extra functionality to a function.

```python
def decorator(func):
    def wrapper(a, b):
        print("Adding numbers...")
        func(a, b)
    return wrapper


@decorator
def add(a, b):
    print(a + b)


add(10, 20)
```

## Closures

### Definition

A closure is a function that remembers variables from its outer function even after the outer function has finished executing.

### Real-Life Example

Imagine your mother gives you a lunch box before you leave for school.

Even after you leave home, you still have the lunch box.

Similarly, the inner function remembers the data from the outer function.

```python
def multiply(x):
    def inner(y):
        print(x * y)

    return inner


result = multiply(5)

result(4)
```

## global and nonlocal

### Definition

`global` allows a function to modify a variable defined outside all functions.

`nonlocal` allows an inner function to modify a variable from its enclosing outer function.

### Real-Life Example

Imagine a company.

- `global` → Company-wide announcement
- `nonlocal` → Department-level announcement

A company announcement affects everyone, while a department announcement affects only that department.

```python
balance = 1000

def deposit():
    global balance
    balance += 500


deposit()

print(balance)
```
