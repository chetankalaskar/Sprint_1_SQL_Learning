# Module 3 – Exception Handling

Exception handling is used to manage errors that occur while a Python program is running.

Instead of allowing the program to stop suddenly, exception handling helps us identify the error and respond to it properly. This can make the program more user-friendly and easier to debug.

The main keywords used in exception handling are:

- try
- except
- else
- finally
- raise
- assert

Python also allows us to create our own exceptions, called custom exceptions.

### What is Exception Handling?

In my understanding, exception handling is a way of dealing with errors that occur while a program is running.

Some errors can cause a program to stop unexpectedly. By using exception handling, we can identify these errors and provide a meaningful message instead of allowing the program to crash.

This also helps the program continue or finish safely.

### Types of Errors

#### 1. Syntax Errors

A syntax error happens when the rules of Python are not followed correctly.

For example, forgetting a colon after an `if` statement can cause a syntax error.

Syntax errors usually occur before the program starts running.

#### 2. Runtime Errors

A runtime error happens while the program is running.

For example, dividing a number by zero causes a runtime error.

Runtime errors can often be handled using exception handling.

### Common Built-in Exceptions

| Exception | Description | Example |
|---|---|---|
| ZeroDivisionError | Raised when dividing a number by zero | `10 / 0` |
| ValueError | Raised when the value is invalid for an operation | `int("abc")` |
| TypeError | Raised when an operation uses incompatible data types | `"10" + 5` |
| NameError | Raised when a variable has not been defined | `print(x)` |
| IndexError | Raised when a list index is out of range | `items[10]` |
| KeyError | Raised when a dictionary key does not exist | `data["age"]` |
| AttributeError | Raised when an object does not have a requested attribute | `100.upper()` |
| FileNotFoundError | Raised when a requested file does not exist | `open("file.txt")` |
| ImportError | Raised when something cannot be imported from a module | `from math import square` |
| ModuleNotFoundError | Raised when a module cannot be found | `import unknown_module` |

## 1. try and except

The `try` block contains code that may produce an error.

The `except` block handles the error if it occurs.

In this example, division by zero can cause an error, so the program handles `ZeroDivisionError`.

````python
try:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))

    result = a / b
    print("Result:", result)

except ZeroDivisionError:
    print("Cannot divide by zero.")
````

## 2. else Block

The `else` block runs only when no exception occurs in the `try` block.

In this example, the program checks whether the entered number is even or odd only if the user enters a valid integer.

````python
try:
    num = int(input("Enter a number: "))

except ValueError:
    print("Please enter a valid integer.")

else:
    if num % 2 == 0:
        print("Even Number")
    else:
        print("Odd Number")
````

## 3. finally Block

The `finally` block runs whether an exception occurs or not.

It is useful for operations that must always be completed, such as closing files or displaying a final message.

````python
try:
    number = 10
    result = number / 2
    print("Result:", result)

except ZeroDivisionError:
    print("Cannot divide by zero.")

finally:
    print("Program completed.")
````

## 8. IndexError

### When does it occur?

`IndexError` occurs when we try to access a position in a list that does not exist.

For example, a list with three items has indexes from 0 to 2.

### Correct way to handle it

We can use `except IndexError` when there is a possibility of accessing an invalid index.

````python
try:
    fruits = ["", "Banana", "Mango"]

    print(fruits[5])

except IndexError:
    print("Invalid list index.")
````

## 9. KeyError

### When does it occur?

`KeyError` occurs when we try to access a dictionary using a key that does not exist.

### Correct way to handle it

We can use `try` and `except KeyError` to handle missing keys.

````python
try:
    marks = {
        "Math": 90,
        "Science": 85,
        "English": 88
    }

    subject = input("Enter subject: ")
    print("Marks:", marks[subject])

except KeyError:
    print("Subject not found.")
````

## 10. AttributeError

### When does it occur?

`AttributeError` occurs when we try to use an attribute or method that does not exist for an object.

For example, the `upper()` method is available for strings but not for integers.

### Correct way to handle it

We can use `except AttributeError`.

````python
try:
    num = 100

    print(num.upper())

except AttributeError:
    print("Integer object has no attribute 'upper'.")
````

## 11. FileNotFoundError

### When does it occur?

`FileNotFoundError` occurs when we try to open a file that does not exist in the specified location.

### Correct way to handle it

We can use `try` and `except FileNotFoundError`.

````python
try:
    file = open("students.txt", "r")

    print(file.read())

    file.close()

except FileNotFoundError:
    print("File not found.")
````

## 12. ImportError

### When does it occur?

`ImportError` can occur when we try to import something that is not available in a module.

For example, the `math` module does not contain a function called `square`.

### Correct way to handle it

We can use `except ImportError`.

````python
try:
    from math import square

except ImportError:
    print("The requested function does not exist in the module.")
````

## 13. ModuleNotFoundError

### When does it occur?

`ModuleNotFoundError` occurs when Python cannot find the module that we are trying to import.

The module may not exist or may not be installed.

### Correct way to handle it

We can use `except ModuleNotFoundError`.

````python
try:
    import my_unknown_module

except ModuleNotFoundError:
    print("Module does not exist or is not installed.")
````

## 14. Using try, except, else and finally

This example demonstrates how the four main blocks work together.

- `try` contains the code that may cause an exception.
- `except` handles specific errors.
- `else` runs when no error occurs.
- `finally` always runs at the end.

````python
try:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))

    result = a / b

except ValueError:
    print("Please enter valid numbers.")

except ZeroDivisionError:
    print("Cannot divide by zero.")

else:
    print("Division successful.")
    print("Result:", result)

finally:
    print("Program completed.")
````

## 15. raise

The `raise` keyword is used when we want to manually create an exception.

We can use it when a specific condition should not be allowed in our program.

In this example, we manually raise a `ZeroDivisionError` when the second number is zero.

````python
try:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))

    if b == 0:
        raise ZeroDivisionError("Cannot divide by zero.")

    result = a / b

except ValueError:
    print("Please enter valid numbers.")

except ZeroDivisionError as error:
    print(error)

else:
    print("Division successful.")
    print("Result:", result)

finally:
    print("Program completed.")
````

## 16. assert

The `assert` statement is used to check whether a condition is true.

If the condition is true, the program continues normally.

If the condition is false, Python raises an `AssertionError`.

In this example, attendance must be between 0 and 100.

````python
try:
    attendance = int(input("Enter attendance percentage: "))

    assert 0 <= attendance <= 100, "Attendance must be between 0 and 100."

    if attendance < 75:
        raise ValueError("Not eligible for the exam.")

except AssertionError as error:
    print(error)

except ValueError as error:
    print(error)

else:
    print("Eligible for the exam.")

finally:
    print("Process completed.")
````

## 17. Custom Exceptions

Python allows us to create our own exceptions according to the requirements of our program.

A custom exception is created by defining a class that inherits from the `Exception` class.

In this example, `PasswordError` is a custom exception. It is raised when the password contains fewer than 8 characters.

````python
class PasswordError(Exception):
    pass


try:
    password = input("Enter password: ")

    if len(password) < 8:
        raise PasswordError(
            "Password must contain at least 8 characters."
        )

except PasswordError as error:
    print(error)

else:
    print("Password accepted.")
````
