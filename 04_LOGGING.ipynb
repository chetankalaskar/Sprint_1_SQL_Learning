## 1. What is Logging?

Logging is the process of recording what happens while a Python program is running.

It can record:

- normal program activities
- warnings
- errors
- important events

Logging is useful because we can check these messages later and understand what happened during program execution.

## 2. Why is Logging used?

Logging is mainly used to:

- debug programs
- track program flow
- record errors
- monitor program behavior
- save information for later analysis

For small programs, `print()` may be enough. But when the program becomes bigger, logging is more useful because we can manage the messages properly.

For example:

```python
print("Chetan")
print("Is working on Module - Logging")
```

`logging` is used to record messages generated by the program.

Logging gives us more control over:

### Main difference

`print()` is useful for simple output and quick checking.

`logging` is more useful for debugging and monitoring real applications.

## 6. Creating a Log File

I can configure logging so that the log messages are stored in a file.

Here:

- `filename` tells where the log file should be created.
- `level` tells which log messages should be recorded.
- `format` decides how the log message should look.

```python
import os
import logging

os.makedirs("logs", exist_ok=True)

logging.basicConfig(
    filename="logs/application.log",
    level=logging.INFO,
    format="%(levelname)s - %(message)s"
)

logging.info("Notebook execution started")

print("Log file created successfully!")
```

## 7. Log Levels

Logging provides different levels depending on the importance of the message.

The main log levels are:

### DEBUG

Used for detailed information while debugging a program.

### INFO

Used for normal information about the program.

### WARNING

Used when something unexpected happens, but the program can still continue.

### ERROR

Used when an error occurs during program execution.

### CRITICAL

Used when a very serious error occurs.

The order of the log levels is:

`DEBUG < INFO < WARNING < ERROR < CRITICAL`

## 8. Using Different Log Levels

We can use different methods from the logging module for different situations.

```python
import logging

logging.basicConfig(level=logging.DEBUG)

logging.debug("Checking values")
logging.info("Task completed")
logging.warning("File is empty")
logging.error("File not found")
logging.critical("Program stopped")
```

### Understanding the messages

- `DEBUG` → detailed information for checking the program
- `INFO` → normal program activity
- `WARNING` → something unexpected happened
- `ERROR` → an operation failed
- `CRITICAL` → a very serious problem occurred

The log level also decides which messages are recorded.

## 9. Formatter

A formatter decides how the log message should look.

For example:

`2026-07-10 15:10:00 | INFO | Application started`

Here we can see:

- date and time
- log level
- actual message

Using a formatter makes the log messages easier to read and understand.

## 10. Creating a Formatter

We can use different format options such as:

- `%(asctime)s` → date and time
- `%(levelname)s` → log level
- `%(message)s` → actual log message

```python
import logging

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s | %(levelname)s | %(message)s"
)

logging.info("Application started")
logging.warning("This is a warning")
logging.error("This is an error")
```

## 11. Handlers

A handler decides where the log message should go.

For example, a log message can be sent to:

- console
- file

There are two handlers that I learned here:

### Console Handler

A console handler displays the log messages on the screen.

### File Handler

A file handler saves the log messages inside a file.

## 12. Using Console and File Handlers

First, I create a logger.

Then I create:

1. a console handler
2. a file handler
3. a formatter

The console handler will display messages on the screen, while the file handler will save them in the log file.

```python
import logging

logger = logging.getLogger("logging_example")
logger.setLevel(logging.DEBUG)

formatter = logging.Formatter(
    "%(asctime)s | %(name)s | %(levelname)s | %(message)s"
)

console_handler = logging.StreamHandler()
console_handler.setLevel(logging.INFO)
console_handler.setFormatter(formatter)

file_handler = logging.FileHandler("logs/application.log")
file_handler.setLevel(logging.DEBUG)
file_handler.setFormatter(formatter)

logger.addHandler(console_handler)
logger.addHandler(file_handler)

logger.debug("Checking values")
logger.info("Task completed")
logger.warning("File is empty")
logger.error("File not found")
```

### What happens here?

The console handler has level `INFO`.

So the `DEBUG` message will not be displayed on the console.

The file handler has level `DEBUG`.

So the file can contain the debug message as well as the other messages.

This means that different handlers can have different log levels.

## 13. Rotating Logs

When a program runs for a long time, the log file can become very large.

To avoid this problem, we can use `RotatingFileHandler`.

It automatically creates backup log files when the main log file reaches a particular size.

### Why is this useful?

It helps to avoid:

- very large log files
- difficult file management
- unnecessary storage usage

## 14. Using RotatingFileHandler

`maxBytes` specifies the maximum size of the log file.

`backupCount` specifies how many backup files should be kept.

```python
from logging.handlers import RotatingFileHandler

rot_logger = logging.getLogger("rotation_demo")
rot_logger.setLevel(logging.DEBUG)

rot_handler = RotatingFileHandler(
    "logs/rotating.log",
    maxBytes=200,
    backupCount=3
)

rot_handler.setFormatter(formatter)
rot_handler.setLevel(logging.DEBUG)

rot_logger.addHandler(rot_handler)

for i in range(1, 25):
    rot_logger.info(f"Log message {i}")

print("Rotating logs created")
```

### Understanding `maxBytes` and `backupCount`

In the above example:

`maxBytes=200`

means the log file is allowed to grow up to around 200 bytes before rotation happens.

`backupCount=3`

means up to 3 backup log files will be kept.

For example, we may get files like:

- `rotating.log`
- `rotating.log.1`
- `rotating.log.2`
- `rotating.log.3`

This prevents one log file from growing continuously.

## 15. Checking if the Log File Exists

We can use `os.path.exists()` to check whether a file exists.

If the file exists, we can open it and read its contents.

```python
import os

if os.path.exists("logs/application.log"):
    print("Log file exists.")

    with open("logs/application.log", "r") as f:
        print(f.read())
else:
    print("Log file not found.")
```
